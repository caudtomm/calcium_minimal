```python
# 0. imports
from pathlib import Path
import numpy as np
from scipy.io import loadmat, savemat
import mat73, h5py, itertools, pandas as pd
from tqdm import tqdm
import time
```

```python
# 1. config
in_dir = Path(r"\\tachyon.fmi.ch\tachyon\groups\scratch\gfriedri\processed-data\250129_TC_invivoCaIMG_odorexp004_005\manifold_data")  # or "/mnt/share/..."
out_dir = Path(r"\\tachyon.fmi.ch\tachyon\groups\scratch\gfriedri\processed-data\250129_TC_invivoCaIMG_odorexp004_005\manifold_data\results")
out_dir.mkdir(parents=True, exist_ok=True)
cell_key = "data"  # set like "C" if you know the variable name; None = auto
save_mode = "mat"  # "mat" or "csv"
csv_float_fmt = "%.9g"

n_repetitions = 150
n_points = 70
```

```python
# 2. external lib
import glue.preprocess
from glue.contrib import glue_analysis_dataframe
def preprocess(manifolds,idx):
    return glue.preprocess.downsample_manifolds(
        manifolds,
        n_points=n_points,
        seed=idx, # use the repetition number as a seed
    ) 
def analyze(manifolds, idx):
    return glue_analysis_dataframe(
      manifolds=manifolds,
      indices=(idx,),
      indices_name=['rep_num'],
      analysis_type='FIRST_VERSUS_REST',
      return_matrix=True,
      shuffle=True,
      gaussianize=False,
      bias=True,
      seed=idx, # use the repetition number as a seed
    )
```

```python
# 3. loaders
def _try_mat73(p):
    try: return mat73.loadmat(p)
    except: return None

def _try_loadmat(p):
    try: return loadmat(p, squeeze_me=True, struct_as_record=False)
    except: return None

def _extract_cell(d, key=None):
    if key and key in d: v = d[key]
    else:
        ks = [k for k in d.keys() if not k.startswith("__")]
        if len(ks)!=1: raise ValueError("ambiguous keys: "+str(ks))
        v = d[ks[0]]
    if isinstance(v, (list, tuple)): return list(v)
    if isinstance(v, np.ndarray) and v.dtype==object:
        return [v.item(i) if v.ndim==1 else v[i] for i in range(v.size)]
    if isinstance(v, np.ndarray) and v.ndim>=2 and v.dtype!=object:
        return [v]
    raise TypeError("not a cell-like object")

def load_cells(p, key=None):
    d = _try_loadmat(p)
    if d is None:
        d = _try_mat73(p)
        if d is None:
            with h5py.File(p,"r") as f: raise RuntimeError("HDF5 v7.3 not parsed; provide key")
    return _extract_cell(d, key)
```

```python
# 4. gather manifolds from all files
mat_files = sorted([p for p in in_dir.glob("*.mat")])
mf = []
src = []
for p in mat_files:
    c = load_cells(p.as_posix(), cell_key)
    for i, arr in enumerate(c):
        a = np.asarray(arr)
        if a.ndim!=2 or a.dtype.kind!='f': a = a.astype(np.float64)
        mf.append(a)
        src.append((p.name, i))

mat_files

print(type(mf),len(mf))
```

```python
# 5. pairwise combos
pairs = list(itertools.combinations(range(len(mf)), 2))
pairs
```

```python
# 6. run and save
start = time.time()

for k,(i,j) in enumerate(tqdm(pairs, desc="pairs")):
    manifolds = [mf[i], mf[j]]
    
    results = []
    outmat_list = []
    for i_rep in range(n_repetitions):
        r1 = preprocess(manifolds,i_rep)
        print(type(r1), [type(x) for x in r1], np.shape(r1[0]), np.shape(r1[1]))

        ret = analyze(r1,i_rep)
        [r2,outmat] = ret
        display(r2)
        type(r2)

        print(type(r2))
        print(type(outmat))

        results.append(r2)
        outmat_list.append(outmat)

    r2 = pd.concat(results)
    display(r2)

    # saving
    base = f"p_{i:05d}_{j:05d}"

    if save_mode == "csv":
        if hasattr(r2, "to_csv"):
            r2.to_csv(out_dir / f"{base}_metrics.csv", index=True, float_format=csv_float_fmt)
        pre_cell = np.empty((1, len(r1)), dtype=object)
        for t, arr in enumerate(r1):
            pre_cell[0, t] = np.asarray(arr)
        savemat(out_dir / f"{base}.mat",
                {
                "preprocessed": pre_cell,
                "alignment": outmat,          # dict -> MATLAB struct
                "src_i": np.array(src[i][0]),
                "idx_i": np.array(src[i][1]),
                "src_j": np.array(src[j][0]),
                "idx_j": np.array(src[j][1]),
                },
                do_compression=True)
    else:
        pre_cell = np.empty((1, len(r1)), dtype=object)
        for t, arr in enumerate(r1):
            pre_cell[0, t] = np.asarray(arr)

        if hasattr(r2, "to_numpy"):
            met_cols = np.array(list(map(str, r2.columns)), dtype=object).reshape(1, -1)  # 1×K cellstr
            if r2.index.dtype == 'object':
                met_idx = np.array(list(map(str, r2.index.tolist())), dtype=object).reshape(-1, 1)
            else:
                met_idx = np.asarray(r2.index.to_numpy()).reshape(-1, 1)
            met_vals = r2.to_numpy()
            metrics_struct = {
                "columns": met_cols,
                "index": met_idx,
                "values": met_vals,
                "index_name": np.array([str(r2.index.name) if r2.index.name is not None else ""], dtype=object),
            }
        else:
            metrics_struct = {"columns": np.array([], dtype=object).reshape(1, 0),
                            "index": np.array([], dtype=float).reshape(0, 1),
                            "values": np.asarray(r2)}

        savemat(out_dir / f"{base}.mat",
                {
                "preprocessed": pre_cell,     # 1×N cell of matrices
                "metrics": metrics_struct,     # struct: columns/index/values
                "alignment": outmat_list,           # dict -> MATLAB struct (no wrapping)
                "src_i": np.array(src[i][0]),
                "idx_i": np.array(src[i][1]),
                "src_j": np.array(src[j][0]),
                "idx_j": np.array(src[j][1]),
                },
                do_compression=True)

print("Time elapsed:", time.time()-start)
```
